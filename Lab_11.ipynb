# Travelling Salesman Problem using Dynamic Programming.
Problem Description:
Given a set of cities and distances, find the shortest possible route that visits each city once and returns to the starting city.

#Theory
TSP uses dynamic programming with bit masking. Store results of visited subsets to reduce repeated work.

## Algorithm
1. Start from source city
2. Mark visited cities using bit mask
3. Try all unvisited cities
4. Choose minimum cost path
5. Return to source city

```python
def tsp(graph):
    n = len(graph)
    dp = [[-1]*(1<<n) for _ in range(n)]
    parent_map = [[-1]*(1<<n) for _ in range(n)] # To store the next city in the optimal path

    def visit(city, mask):
        if mask == (1<<n) - 1:
            # If all cities visited, return cost to go back to the starting city (city 0)
            return graph[city][0]

        if dp[city][mask] != -1:
            return dp[city][mask]

        min_ans = float('inf')
        best_next_city = -1

        for next_city in range(n):
            # Check if next_city is not yet visited
            if not (mask & (1<<next_city)):
                cost_to_next = graph[city][next_city]
                # Ensure there's a valid path (cost > 0 or it's a self-loop of 0 when city==next_city is not the current step)
                if cost_to_next == 0 and city != next_city and next_city != 0: # Avoid path through 0 with 0 cost unless it's the destination
                    continue

                new_cost = cost_to_next + visit(next_city, mask | (1<<next_city))
                if new_cost < min_ans:
                    min_ans = new_cost
                    best_next_city = next_city

        dp[city][mask] = min_ans
        parent_map[city][mask] = best_next_city # Store the chosen next city for path reconstruction
        return min_ans

    # Start TSP from city 0 with only city 0 visited in the mask
    min_total_cost = visit(0, 1)

    # Reconstruct the path
    path = []
    current_city = 0
    current_mask = 1 # Represents that city 0 has been visited

    while current_mask != (1 << n) - 1:
        path.append(current_city)
        next_city = parent_map[current_city][current_mask]
        if next_city == -1: # Should not happen if path exists, but as a safeguard
            break
        current_city = next_city
        current_mask |= (1 << current_city) # Add the next_city to the mask

    path.append(current_city) # Add the last visited city before returning to start
    path.append(0) # Complete the cycle by returning to the starting city

    return min_total_cost, path
```

```python
graph = [
[0, 10, 15, 20],
[10, 0, 35, 25],
[15, 35, 0, 30],
[20, 25, 30, 0]
]

print("Minimum travelling cost:", tsp(graph))
```

## User Input and Testing

```python
# Helper function to parse user input for the graph
def get_user_graph():
    while True:
        try:
            n = int(input("Enter the number of cities: "))
            if n < 2:
                print("Number of cities must be at least 2. Please try again.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter an integer.")

    graph = []
    print(f"Enter the {n}x{n} adjacency matrix row by row. Use spaces to separate values.")
    print("Example for 3 cities: 0 10 15, then new line, 10 0 20, then new line, 15 20 0")

    for i in range(n):
        while True:
            try:
                row_str = input(f"Enter row {i+1} (space-separated integers): ")
                row = list(map(int, row_str.split()))
                if len(row) != n:
                    print(f"Row must contain {n} values. Please re-enter row {i+1}.")
                    continue
                graph.append(row)
                break
            except ValueError:
                print("Invalid input. Please enter space-separated integers.")
    return graph

# Test with user-provided graph
print("--- Testing with User Input ---")
user_graph = get_user_graph()
min_cost, path = tsp(user_graph)
print(f"User graph: {user_graph}")
print(f"Minimum travelling cost for user graph: {min_cost}")
print(f"Optimal path for user graph: {path}")
```

```python
# Test with different numbers of cities
print("\n--- Testing with Different Number of Cities ---")

# 3 Cities
print("\n--- 3 Cities Example ---")
graph_3_cities = [
    [0, 10, 15],
    [10, 0, 20],
    [15, 20, 0]
]
min_cost_3, path_3 = tsp(graph_3_cities)
print(f"Graph (3 cities): {graph_3_cities}")
print(f"Minimum travelling cost (3 cities): {min_cost_3}")
print(f"Optimal path (3 cities): {path_3}")

# 5 Cities
print("\n--- 5 Cities Example ---")
graph_5_cities = [
    [0, 20, 42, 35, 25],
    [20, 0, 30, 34, 12],
    [42, 30, 0, 10, 6],
    [35, 34, 10, 0, 15],
    [25, 12, 6, 15, 0]
]
min_cost_5, path_5 = tsp(graph_5_cities)
print(f"Graph (5 cities): {graph_5_cities}")
print(f"Minimum travelling cost (5 cities): {min_cost_5}")
print(f"Optimal path (5 cities): {path_5}")
```
